Q1 - resposta
Ao realizar a exploração inicial dos dados, foi possível observar que o DataFrame possui informações sobre imóveis, como área, quantidade de quartos, banheiros, vagas de estacionamento e valores financeiros.
as colunas financeiras, como `hoa`, `rent amount`, `property tax`, `fire insurance` e `total`, estão inicialmente armazenadas como texto `object`.
também é possível observar que algumas colunas categóricas, como `animal` e `furniture`, estão armazenadas como texto.
e a verificação de valores nulos permite identificar se existem registros sem informações. Caso alguma coluna apresente valores nulos, eles deverão ser tratados antes da utilização do modelo de regressão.

```python
#2a
colunas_financeiras = ['hoa', 'rent amount', 'property tax', 'fire insurance', 'total']

for coluna in colunas_financeiras:
    df_aluguel[coluna] = df_aluguel[coluna].str.replace('R$', '', regex=False)
    df_aluguel[coluna] = df_aluguel[coluna].str.replace(',', '', regex=False)
    df_aluguel[coluna] = df_aluguel[coluna].replace('Sem info', '0')
    df_aluguel[coluna] = df_aluguel[coluna].replace('Incluso', '0')
    df_aluguel[coluna] = pd.to_numeric(df_aluguel[coluna], errors='coerce')
    df_aluguel[coluna] = df_aluguel[coluna].fillna(0)
```

```python
#2b
df_aluguel['floor'] = df_aluguel['floor'].replace('-', 0)
df_aluguel['floor'] = df_aluguel['floor'].astype(int)
df_aluguel['floor'].head()
```

```python
#2c
df_aluguel['animal'] = df_aluguel['animal'].map({'acept': 1, 'not acept': 0})
df_aluguel['furniture'] = df_aluguel['furniture'].map({'furnished': 1, 'not furnished': 0})
df_aluguel[['animal', 'furniture']].head()
```

```python
df_aluguel.info()
```

```python
#3
y = df_aluguel['total']
X = df_aluguel.drop('total', axis=1)
modelo = LinearRegression()
modelo.fit(X, y)                 
print('Intercepto (valor base):', modelo.intercept_)
print('Coeficientes de cada variável:')
for nome, valor in zip(X.columns, modelo.coef_):
    print(f"{nome}: {valor:.2f}")
```

```python
#4
dados_exemplo = pd.DataFrame([{'city': 1, 'area': 80, 'rooms': 3, 'bathroom': 2, 
                               'parking spaces': 1, 'floor': 5, 
                               'animal': 1, 'furniture': 1, 'hoa': 800, 
                               'rent amount': 3000, 'property tax': 150, 
                               'fire insurance': 30}])

previsao = modelo.predict(dados_exemplo)
print("Valor total previsto:", previsao[0])
```

Q4 - resposta
Com base nos coeficientes calculados pelo modelo, a variável que apresentou o maior impacto foi 'rent amount', pois apresentou o maior valor absoluto de coeficiente.
O valor absoluto foi utilizado para considerar tanto impactos positivos quanto negativos. Um coeficiente positivo indica uma relação estimada de aumento no valor total, enquanto um coeficiente negativo indica uma relação estimada de redução no valor total, mantendo as demais variáveis constantes.
